### **Task 2c. Understand the data - Investigate the variables**

In this phase, you will begin to investigate the variables more closely to better understand them.

You know from the project proposal that the ultimate objective is to use machine learning to classify videos as either claims or opinions. A good first step towards understanding the data might therefore be examining the `claim_status` variable. Begin by determining how many videos there are for each different claim status.

```python
# What are the different values for claim status and how many of each are in the data?
data['claim_status'].value_counts(dropna=False)
```

**Question:** What do you notice about the values shown?

**Answer:**
- The target variable `claim_status` has two distinct classes: **`claim` (9,608 records)** and **`opinion` (9,476 records)**, alongside **298 missing (`NaN`) records**.
- Among non-null records, the classes are virtually evenly balanced: **50.3% claims vs 49.7% opinions**.
- This balanced distribution is an optimal condition for training classification models, as it prevents algorithmic bias toward the majority class without requiring artificial resampling (e.g., SMOTE or undersampling).

Next, examine the engagement trends associated with each different claim status.

Start by using Boolean masking to filter the data according to claim status, then calculate the mean and median view counts for each claim status.

```python
# What is the average view count of videos with "claim" status?
print('Mean view count (claim):', data[data['claim_status'] == 'claim']['video_view_count'].mean())
print('Median view count (claim):', data[data['claim_status'] == 'claim']['video_view_count'].median())
```

```python
# What is the average view count of videos with "opinion" status?
print('Mean view count (opinion):', data[data['claim_status'] == 'opinion']['video_view_count'].mean())
print('Median view count (opinion):', data[data['claim_status'] == 'opinion']['video_view_count'].median())
```

**Question:** What do you notice about the mean and median within each claim category?

**Answer:**
- **Extreme Scale Disparity:** Videos classified as `claim` receive an average of **501,029.45 views** (median: **501,555.00**), whereas videos classified as `opinion` receive an average of only **4,956.43 views** (median: **4,953.00**). Videos with factual claims receive roughly **100 times more views** than opinion videos.
- **Symmetry within Categories:** Within each individual category, the mean and median are remarkably close (~501k vs ~501.5k for claims; ~4.95k vs ~4.95k for opinions). This demonstrates that within each class, view counts are uniformly distributed across their respective operational ranges, and the overall bimodal skew of the dataset is driven directly by `claim_status`.

Now, examine trends associated with the ban status of the author.

Use `groupby()` to calculate how many videos there are for each combination of categories of claim status and author ban status.

```python
# Get counts for each group combination of claim status and author ban status
data.groupby(['claim_status', 'author_ban_status']).count()[['#']]
```

**Question:** What do you notice about the number of claims videos with banned authors? Why might this relationship occur?

**Answer:**
- **Overwhelming Concentration in Claims:** Of the 1,635 videos published by banned authors, **1,439 (88.0%) are claims**, while only 196 (12.0%) are opinions. Similarly, for authors under review / under scrutiny, **1,603 (77.6%) are claims** and only 463 (22.4%) are opinions. For active authors, opinion videos actually outnumber claims (8,817 opinions vs 6,566 claims).
- **Underlying Cause:** Claims assert factual statements that can be verified or falsified. Videos containing false claims, medical misinformation, conspiracy theories, or misleading factual assertions directly breach TikTok Community Guidelines, triggering user flags, moderator review, and disciplinary sanctions (scrutiny or account bans). Personal opinions, by contrast, are protected speech under most platform policies and rarely incur bans.

Continue investigating engagement levels, now focusing on `author_ban_status`.

Calculate the median video share count of each author ban status.

```python
# Calculate the median video share count of each author ban status
data.groupby('author_ban_status')['video_share_count'].median()
```

```python
# What's the median video share count of each author ban status?
data.groupby('author_ban_status')[['video_share_count']].agg('median')
```

**Question:** What do you notice about the share count of banned authors, compared to that of active authors? Explore this in more depth.

**Answer:**
- The median share count for videos from banned authors is **14,468.0**, and for authors under review it is **9,444.0**, compared to just **437.0** for active authors.
- Videos from banned authors are shared **over 33 times more often** than videos from active authors. This highlights that content violating community guidelines often achieves rapid virality and wide dissemination before enforcement actions take place.

Use `groupby()` to group the data by `author_ban_status`, then use `agg()` to get the count, mean, and median of each of the following columns:
* `video_view_count`
* `video_like_count`
* `video_share_count`

Remember, the argument for the `agg()` function is a dictionary whose keys are columns. The values for each column are a list of the calculations you want to perform.

```python
# Group by author_ban_status and calculate count, mean, and median for views, likes, and shares
data.groupby('author_ban_status').agg({
    'video_view_count': ['count', 'mean', 'median'],
    'video_like_count': ['count', 'mean', 'median'],
    'video_share_count': ['count', 'mean', 'median']
})
```

**Question:** What do you notice about the number of views, likes, and shares for banned authors compared to active authors?

**Answer:**
- Across all three engagement metrics (views, likes, shares), **banned authors and authors under review consistently generate far higher volumes** than active authors:
  - **Mean Views:** Banned authors average **445,845.4 views** (median: 448,201) and authors under review average **392,204.8 views** (median: 379,440.5), compared to active authors who average **215,927.0 views** (median: 8,616).
  - **Mean Likes:** Banned authors average **153,017.4 likes** (median: 105,573) vs **71,034.4 likes** (median: 2,222) for active authors.
  - **Mean Shares:** Banned authors average **29,998.9 shares** (median: 14,468) vs **14,111.5 shares** (median: 437) for active authors.
- This reinforces the critical finding that violating or sensationalist content spreads vigorously, driving heightened user engagement before moderator intervention.

Now, create three new columns to help better understand engagement rates:
* `likes_per_view`: represents the number of likes divided by the number of views for each video
* `comments_per_view`: represents the number of comments divided by the number of views for each video
* `shares_per_view`: represents the number of shares divided by the number of views for each video

```python
# Create a likes_per_view column
data['likes_per_view'] = data['video_like_count'] / data['video_view_count']

# Create a comments_per_view column
data['comments_per_view'] = data['video_comment_count'] / data['video_view_count']

# Create a shares_per_view column
data['shares_per_view'] = data['video_share_count'] / data['video_view_count']
```

Use `groupby()` to compile the information in each of the three newly created columns for each combination of categories of claim status and author ban status, then use `agg()` to calculate the count, the mean, and the median of each group.

```python
# Compile information for rate columns across claim status and author ban status
data.groupby(['claim_status', 'author_ban_status']).agg({
    'likes_per_view': ['count', 'mean', 'median'],
    'comments_per_view': ['count', 'mean', 'median'],
    'shares_per_view': ['count', 'mean', 'median']
})
```

**Question:**

How does the data for claim videos and opinion videos compare or differ? Consider views, comments, likes, and shares.

**Answer:**
- **Volume Disparity:** Claim videos receive substantially higher raw engagement across all dimensions (~501k average views vs ~4.9k for opinions; ~166k likes vs ~1.7k; ~33k shares vs ~330).
- **Engagement Rate Disparity:** Beyond raw volume, claim videos exhibit consistently higher engagement intensity per view:
  - `likes_per_view`: Claim videos convert viewers to likes at a rate of **~32.8% to 34.5%**, whereas opinion videos convert at **~20.7% to 22.6%** (~50% relative increase).
  - `comments_per_view`: Claim videos generate **~7.0%** comments per view, whereas opinions generate **~4.6%** (~52% relative increase).
  - `shares_per_view`: Claim videos produce **~6.5% to 6.8%** shares per view, whereas opinions produce **~4.1% to 4.4%** (~55% relative increase).
- **Consistency Across Ban Statuses:** These elevated engagement rates for claims persist whether the creator is active, under review, or banned. Factual claims inherently stimulate higher interaction, discussion, and social dissemination than personal opinions.
